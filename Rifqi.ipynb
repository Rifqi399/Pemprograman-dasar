```python
#Rifqi Al Ghifari Rambe
#4232501017
# 26
# Untuk menghitung efisiensi termal suatu unit pembangkit berdasarkan energi input dan energi output menggunakan rumus matematika
#serta memahami penggunaan operator aritmatika, tanda kurung, dan f-string dalam Python.


energi_input = float(input("memasukan energi input(MJ) : "))
energi_output = float(input("memasukan energi output (MJ)"))
efisiensi = (energi_input/ energi_output) * 100
print(f"Efisiensi termal: {efisiensi:.2f}%")
```

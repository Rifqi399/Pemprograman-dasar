Your own array

Make a list temps with 7 daily temperatures (any values). Then: (1) show the first 3 days and the last 2 days with slicing; (2) add one day with append, then take it out again with pop; (3) show a sorted version with sorted without changing temps; (4) compute the average and find the highest temperature and which day it was; (5) build a new list of the temperatures above the average.

````python

# Membuat list suhu selama 7 hari
temps = [22, 24, 21, 26, 28, 25, 23]

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("First 3 days:", temps[:3])
print("Last 2 days:", temps[-2:])

# 2. Menambahkan dan menghapus satu suhu
temps.append(30)
print("After append:", temps)

temps.pop()
print("After pop:", temps)

# 3. Mengurutkan tanpa mengubah list asli
print("Sorted temperatures:", sorted(temps))

# 4. Menghitung rata-rata dan suhu tertinggi
average = sum(temps) / len(temps)
highest = max(temps)
day = temps.index(highest) + 1

print("Average:", average)
print("Highest temperature:", highest)
print("Highest temperature day:", day)

# 5. Membuat list suhu di atas rata-rata
above_average = [t for t in temps if t > average]

print("Above average:", above_average)
````

Warm-up

Five small exercises (tiny data)

Use scores = [80, 65, 90, 75, 70, 85]. Write code that (1) prints the first score and the last score (use a negative index for the last one), (2) prints the average, (3) counts how many scores are above the average, (4) prints the position of the lowest score as a student number (first student = 1), and (5) builds a new list with every score increased by 5.

````python

# Warm-up: Five Small Exercises

scores = [80, 65, 90, 75, 70, 85]

# 1. Print first and last score
print("First score:", scores[0])
print("Last score:", scores[-1])

# 2. Calculate average
average = sum(scores) / len(scores)
print("Average:", average)

# 3. Count scores above average
above = 0

for score in scores:
    if score > average:
        above += 1

print("Scores above average:", above)

# 4. Find position of lowest score
lowest = min(scores)
position = scores.index(lowest) + 1

print("Lowest score:", lowest)
print("Student number:", position)

# 5. Create a new list with every score increased by 5
new_scores = []

for score in scores:
    new_scores.append(score + 5)

print("New scores:", new_scores)
````

Lab exercise

Monthly production report (portfolio)

````python

# Monthly Production Report

# Data energi selama 30 hari
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

days = list(range(1, 31))

# 1. Total dan rata-rata energi
total = sum(energy)
average = total / len(energy)

print("Total energy:", total, "MWh")
print("Average energy:", round(average, 2), "MWh")

# 2. Tiga hari dengan energi tertinggi
highest = sorted(enumerate(energy, 1),
                 key=lambda x: x[1], reverse=True)[:3]

print("\nThree highest days:")
for day, value in highest:
    print("Day", day, ":", value, "MWh")

# Tiga hari dengan energi terendah
lowest = sorted(enumerate(energy, 1),
                key=lambda x: x[1])[:3]

print("\nThree lowest days:")
for day, value in lowest:
    print("Day", day, ":", value, "MWh")

# 3. Menghitung hari di bawah target
target = float(input("\nEnter energy target: "))

below_target = [e for e in energy if e < target]

print("Days below target:", len(below_target))

# 4. Menghitung 7-day moving average
limit = average * 0.97

print("\nDays with 7-day average below 97%:")

for i in range(6, len(energy)):
    moving_average = sum(energy[i-6:i+1]) / 7

    if moving_average < limit:
        print("Day", days[i],
              ":", round(moving_average, 2), "MWh")

# 5. Total energi per minggu
print("\nWeekly energy totals:")

for week in range(4):
    start = week * 7
    end = start + 7
    weekly_total = sum(energy[start:end])

    print("Week", week + 1, ":", weekly_total, "MWh")

# Sisa hari (hari ke-29 dan 30)
remaining = sum(energy[28:])

print("Remaining days:", remaining, "MWh")
````

```python
tahun = int(input("memasukan_tahun: "))
kabisat = (tahun % 4 == 0 and tahun % 100 !=0) or (tahun % 400 == 0)

print("tahun_kabisat:" , kabisat)
```

Penjelasan singkat tugas 1 = Tahun kabisat adalah tahun yang memiliki 366 hari bukan 365 dan tahun kabisat ini biasanya ditandakan seperti bulan febuary memeliki 29 hari, tahun habis dibagi 4 tapi tidak habis dibagi 100 dan tahun habis dibagi 400

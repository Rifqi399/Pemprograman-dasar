Soal 1. Daya Generator dan Efisiensi Turbin (10 poin | CLO-1 | TP-1, TP-3)

```python
tegangan = 220
arus = 15

energi_input = 500
energi_output = 400

# Menghitung daya listrik
daya = tegangan * arus

# Menghitung efisiensi turbin
efisiensi = (energi_output / energi_input) * 100

# Menampilkan hasil
print(f"Daya listrik = {daya} W")
print(f"Efisiensi = {efisiensi:.2f}%")
```

Soal 9. Pembacaan Vibrasi dengan Sensor Rusak (30 poin | CLO-1 | TP-5, TP-6)

```python

# Data vibrasi dari lima unit mesin
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

# Variabel untuk menghitung data tidak valid
jumlah_tidak_valid = 0

# Memeriksa setiap data menggunakan perulangan for
for i in range(len(data)):
    try:
        # Mengubah data teks menjadi float
        vibrasi = float(data[i])

        # Memeriksa apakah vibrasi melebihi 2.0
        if vibrasi > 2.0:
            print(f"Warning: Unit {i+1} vibrasi tinggi!")

            # Jika vibrasi melebihi 3.0, hentikan pemeriksaan
            if vibrasi > 3.0:
                print("KRITIS: pemeriksaan dihentikan")
                break

    except ValueError:
        # Jika data tidak bisa diubah menjadi float
        print(f"Unit {i+1}: data tidak valid")
        jumlah_tidak_valid += 1

# Menampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

Challenge: Bonus Nilai (hingga +25 poin) (CLO-1 | TP-4, TP-5, TP-6)

```python

# PROGRAM PEMANTAUAN KONSUMSI ENERGI LISTRIK HARIAN

# List untuk menyimpan data konsumsi listrik selama 7 hari
konsumsi = [12.5, 15.0, 18.5, 22.0, 30.5, 17.0, 14.5]

# Variabel untuk menghitung total konsumsi listrik
total = 0

# Menampilkan judul program
print("=== PEMANTAUAN KONSUMSI ENERGI LISTRIK ===")

# Perulangan untuk memeriksa konsumsi setiap hari
for i in range(len(konsumsi)):
    try:
        # Mengambil data konsumsi listrik
        energi = float(konsumsi[i])

        # Menambahkan konsumsi ke total
        total += energi

        # Menampilkan konsumsi setiap hari
        print(f"Hari ke-{i+1}: {energi} kWh")

        # Percabangan untuk menentukan kondisi energi
        if energi <= 15:
            print("Status: Hemat energi")

        elif energi <= 25:
            print("Status: Konsumsi normal")

        else:
            print("Status: Boros energi!")

    except ValueError:
        print(f"Data hari ke-{i+1} tidak valid")

# Menghitung rata-rata konsumsi listrik
rata_rata = total / len(konsumsi)

# Menampilkan hasil perhitungan
print("\n=== HASIL PEMANTAUAN ===")
print(f"Total konsumsi listrik: {total:.2f} kWh")
print(f"Rata-rata konsumsi: {rata_rata:.2f} kWh per hari")

# Menentukan kondisi konsumsi rata-rata
if rata_rata <= 15:
    print("Kesimpulan: Penggunaan listrik sangat hemat.")

elif rata_rata <= 25:
    print("Kesimpulan: Penggunaan listrik dalam kondisi normal.")

else:
    print("Kesimpulan: Penggunaan listrik terlalu tinggi.")
```
